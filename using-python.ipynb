# Analysis on Diamonds Dataset

### Importing Libraries

We use **pandas** for data manipulation, **matplotlib** for plotting, and **seaborn** for statistical visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

### Loading Dataset

The diamonds dataset is included with seaborn — the same dataset used in R's ggplot2 package. We load it directly without needing a local CSV file.

```python
diamonds_df = sns.load_dataset("diamonds")

diamonds_df.head()
```

### Boxplot

A boxplot shows how diamond **price** is distributed across different **cut** quality levels. Each box displays the median, quartiles, and potential outliers.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=diamonds_df, x="cut", y="price", hue="cut", palette="Set2", legend=False)
plt.title("Diamond Price by Cut Quality")
plt.xlabel("Cut")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

### Histograms

We can plot histograms for each level of the **cut** variable using faceted subplots. This lets us compare the price distribution within each cut category side by side.

```python
g = sns.displot(
    data=diamonds_df,
    x="price",
    col="cut",
    col_wrap=1,
    bins=20,
    height=3,
    aspect=3,
    facet_kws={"sharex": False, "sharey": True},
)
g.set_axis_labels("Price", "Count")
g.set_titles("Cut: {col_name}")
plt.tight_layout()
g.savefig("02_histogram_diamonds_cut.png")
plt.show()
```

### Multivariate Analysis

We select numeric columns and compute the **correlation matrix** to understand linear relationships between carat, depth, price, and table.

```python
keep_vars = ["carat", "depth", "price", "table"]
df = diamonds_df[keep_vars]

M_cor = df.corr()
M_cor
```

Showing it in a **heat map** makes it easier to spot strong positive or negative correlations at a glance.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(M_cor, annot=True, cmap="coolwarm", vmin=-1, vmax=1, square=True)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### Price and Carat Correlation

A **pair plot** (Python's equivalent of R's `ggpairs`) shows scatter plots, histograms, and correlation coefficients for all pairs of numeric variables.

```python
sns.pairplot(df, height=2.5, aspect=1.5)
plt.show()
```
