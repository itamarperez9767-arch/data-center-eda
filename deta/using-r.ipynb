# Analysis on Diamonds Dataset

### ⭐ Importing Libraries

```r
# It contains all the R packages that one needs to perfrom data analytics
library(tidyverse)
library(GGally)
```

### ⭐ Loading dataset

```r
# Loading dataset
data("diamonds")

head(diamonds)
```

### ⭐ Boxplot 

```r
p = ggplot(diamonds, aes(x = cut, y = price, fill = cut)) + geom_boxplot() + theme_bw()
print(p)
```

### ⭐ Histograms
*We can plot histograms for each level of the cut factor variable using facet_grid*

```r
# We can plot histograms for each level of the cut factor variable using facet_grid

options(repr.plot.width = 10, repr.plot.height = 8)
p = ggplot(diamonds, aes(x = price, fill = cut)) + geom_histogram(bins=20) + facet_grid(cut ~ ., scales = "free") + theme_bw() 
p  
png('02_histogram_diamonds_cut.png') 
print(p) 
dev.off()
```

### ⭐ MultiVariative analysis

```r
# Correlation matrix plots  
keep_vars = c('carat', 'depth', 'price', 'table') 
df = diamonds[, keep_vars]  
# compute the correlation matrix 
M_cor = cor(df) 
M_cor
```

*Showing it in Heat Map*

```r
heatmap(M_cor)
```

### ⭐ Price and Carat correlation

```r
options(repr.plot.width=15, repr.plot.height=8)
ggpairs(df)
```
